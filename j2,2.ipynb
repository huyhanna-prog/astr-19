```
import matplotlib.pyplot as plt
import numpy as np
```

```
x=np.linspace(0, 1, 100)
def sin_x(val):
    return np.sin(val)
def cos_x(val):
    return np.cos(val)
```

```
fig, (ax1, ax2)= plt.subplots(1, 2, figsize=(10, 4))
ax1.plot(x, sin_x(x))
ax1.set_title('sin')
ax2.plot(x, cos_x(x))
ax2.set_title('cos')
plt.savefig('fig.pdf')
```
