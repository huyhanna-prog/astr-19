```
import matplotlib.pyplot as plt
import numpy as np
```

```
x=np.linspace(0, 1, 100)
```

```
def exp_func(arr):
    return np.exp(arr)
```

```
y=exp_func(x)
plt.plot(x,y)
plt.xlabel('Time in milliseconds')
plt.ylabel('awesomeness')
plt.savefig('awesome.pdf')
```
