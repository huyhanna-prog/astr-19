```
import matplotlib.pyplot as plt
import numpy as np
```

```
n= 100
p= .5
size= 1000

random_numbers = np.random.binomial(n, p, size)
plt.hist(random_numbers, bins= 100)

plt.xlabel('x axis')
plt.ylabel('y axis')

plt.savefig('binomial.pdf')
```
